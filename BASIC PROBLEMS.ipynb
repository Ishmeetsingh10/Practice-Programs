```python
#Program to print sum of all the digits of given number
n = int(input("Enter a number: "))

sum = 0

while n > 0:
    digit = n % 10
    sum = sum + digit
    n = n // 10

print("Sum of digits:", sum)
```

```python
#find the factorial of a given number
n = int(input("Enter a number: "))

fact = 1

for i in range(1, n + 1):
    fact = fact * i

print("Factorial:", fact)
```

```python
#python program to check if given number prime or not
n = int(input("Enter a number: "))

count = 0

for i in range(1, n + 1):
    if n % i == 0:
        count = count + 1

if count == 2:
    print("Prime")
else:
    print("Not Prime")
```

```python
#python program to print prime numbers between 10 to 99
for n in range(10, 100):
    count = 0

    for i in range(1, n + 1):
        if n % i == 0:
            count = count + 1

    if count == 2:
        print(n)
```

```python
#Program to reverse a given number
n = int(input("Enter a number: "))

reverse = 0

while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n = n // 10

print("Reverse:", reverse)
```

```python
#To determine whether a given year is a leap year or not.
year = int(input("Enter year: "))

if year % 400 == 0:
    print("Leap Year")
elif year % 100 == 0:
    print("Not a Leap Year")
elif year % 4 == 0:
    print("Leap Year")
else:
    print("Not a Leap Year")
```

```python
#In certain university, grades are assigned to the marks obtained in respective subjects.
#Grades are assigned as given below.
#Write a python program to accept Marks in Math and display respective grade as an output using if…elif…elif..else statements
#Hint : Relational operators: <, >, <=, >=, ==, !=

#Marks obtained    Grade assigned
#90-100           O
#80-89            A+
#70-79            A
#60-69            B
#50-59            C
#40-49            P
#Below 39         F

marks = int(input("Enter marks: "))

if marks >= 90 and marks <= 100:
    print("Grade: O")
elif marks >= 80:
    print("Grade: A+")
elif marks >= 70:
    print("Grade: A")
elif marks >= 60:
    print("Grade: B")
elif marks >= 50:
    print("Grade: C")
elif marks >= 40:
    print("Grade: P")
else:
    print("Grade: F")

```

```python
#Accept Student  Name, Roll Number and Marks of the 3 subjects from the user.
#Calculate the percentage of the marks and display it. Display the Subject with Highest and lowest marks.

name = input("Enter student name: ")
roll = int(input("Enter roll number: "))

math = int(input("Enter Maths marks: "))
python = int(input("Enter Python marks: "))
english = int(input("Enter English marks: "))

total = math + python + english
percentage = total / 3

print("Student Name:", name)
print("Roll Number:", roll)
print("Percentage:", percentage)

if math > python and math > english:
    print("Highest marks: Maths")
elif python > math and python > english:
    print("Highest marks: Python")
else:
    print("Highest marks: English")

if math < python and math < english:
    print("Lowest marks: Maths")
elif python < math and python < english:
    print("Lowest marks: Python")
else:
    print("Lowest marks: English")
```
