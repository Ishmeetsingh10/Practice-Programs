```python
#write a program to iterate over a dictionary using for loop and print the keys alone,values alone and both keys and values.

student = {
    "1": "a",
    "2": "b",
    "3": "c",
    "4": "d",
}

print("Keys:")
for key in student:
    print(key)

print("\nValues:")
for value in student.values():
    print(value)

print("\nKeys and Values:")
for key, value in student.items():
    print(key, ":", value)
```

```python
#write a python program to create a dictionary of 5 students with marks and display the students with the highest marks
students = {
    "Amit": 85,
    "Riya": 92,
    "Rahul": 78,
    "Sneha": 95,
    "Karan": 88
}
highest_student=max(students, key=students.get)
print ("student with highest marks:", highest_student)
print ("highest marks:", students[ highest_student])
```

```python
#python program to count the frequency of each element in a list using dictionary    
numbers = [10,20,10,30,20,10,40,30]
frequency = {}
for num in numbers:
    if num in frequency:
        frequency[num]= frequency[num]+1
    else:
        frequency[num] = 1
print("Frequency of elements:",frequency)    
```

```python
#python program to sort a dictionary by its values
students = {
    "Amit": 85,
    "Rahul": 92,
    "Priya": 88,
    "Sneha": 95
}

sorted_students = dict(sorted(students.items(), key=lambda item: item[1]))

print("Dictionary sorted by values:")
print(sorted_students)

```

```python
#python program to sort a dictionary by its values   #reverse
students = {
    "Amit": 85,
    "Rahul": 92,
    "Priya": 88,
    "Sneha": 95
}

sorted_students = dict(sorted(students.items(), key=lambda item: item[1],reverse=True))

print("Dictionary sorted by values:")
print(sorted_students)
```
