```python
#accept 20 values from user and save it in list 
#a count similar elements of list and print their index value
numbers = []

for i in range(20):
    x = int(input("Enter value: "))
    numbers.append(x)

for i in range(20):
    count = numbers.count(numbers[i])
    
    if count > 1:
        print(numbers[i], "occurs", count, "times")
        print("Index:", i)
```

```python
#Accept 10 values from user and save it in list. Perform following operations on it  
#a) Sort list in ascending order using sorted() function and display sorted list
#b) sort list in descending order using sort() function
#c) display length of list
numbers = []

for i in range(10):
    x = int(input("Enter value: "))
    numbers.append(x)

# a) sorted() ascending
ascending = sorted(numbers)
print("Ascending:", ascending)

# b) sort() descending
numbers.sort(reverse=True)
print("Descending:", numbers)

# c) length
print("Length:", len(numbers))
```

```python
# Accept two lists from user and merge them using + in a single list
list1 = []

n1 = int(input("How many elements in list1? "))

for i in range(n1):
    value = int(input("Enter value: "))
    list1.append(value)

list2 = []

n2 = int(input("How many elements in list2? "))

for i in range(n2):
    value = int(input("Enter value: "))
    list2.append(value)

list3 = list1 + list2

print("Merged list:", list3)
```

```python
#An acronym is a word formed by taking the first letters of the words in a phrase and making a word from them.
#For example, RAM is an acronym for “random access memory.” 
#Write a program that allows the user to type in a phrase and then outputs the acronym for that phrase.
#Note: the acronym should be all uppercase, even if the words in the phrase are not capitalized.

phrase = input("Enter a phrase: ")

words = phrase.split()

acronym = ""

for word in words:
    acronym = acronym + word[0]

print("Acronym:", acronym.upper())
```

```python
#Write a program to print the abbreviation of a month, given its number
month = int(input("Enter month number: "))

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

if 1 <= month <= 12:
    print("Month abbreviation:", months[month - 1])
else:
    print("Invalid month number")
```

```python
#Python program to merge two lists into a dictionary (one list as keys, another as values).
keys = ["name", "age", "city"]
values = ["ishmeet", 21, "Pune"]

dictionary = dict(zip(keys, values))

print(dictionary)
```

```python
#Python program to remove duplicate elements from a list.
numbers = [10, 20, 10, 30, 20, 40, 30]

unique = []

for n in numbers:
    if n not in unique:
        unique.append(n)

print("List after removing duplicates:", unique)
```

```python
#Python program to convert two lists into a dictionary using zip().
keys = ["A", "B", "C"]
values = [100, 200, 300]

dictionary = dict(zip(keys, values))

print("Dictionary:", dictionary)
```

```python
#Python program to find common elements in two lists.
list1 = [10, 20, 30, 40, 50]
list2 = [30, 40, 50, 60, 70]

common = []

for n in list1:
    if n in list2:
        common.append(n)

print("Common elements:", common)
```
