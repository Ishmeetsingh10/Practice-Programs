```python
#write a program that accepts the three numbers from user and 
#display its addition using function

x = int(input("enter X"))
y = int(input("enter Y"))
z = int(input("enter Z"))

def cal_sum(x,y,z):
    sum = x+y+z
    return sum

s=cal_sum(x,y,z)    
print(s)
```

```python
#sum and productt
def cal_sum_prod(x,y,z):
    ss = x+y+z
    pp = x*y*z
    return ss,pp

a = int(input("enter a"))
b = int(input("enter b"))
c = int(input("enter c"))
s,p = cal_sum_prod(a,b,c)
print(s,p)
```

```python
#accept the number from the user and display prime factors of that no.

def prime_factors(n):
    for i in range (2, n+1):
        if n%i == 0:
            print(i,end=" ")
            while n%i == 0:
                n = n // i

num = int(input("enter number:"))            
prime_factors(num)
```

```python
#which will print all even nos from a range 1 to 100
def even_numbers():
    for i in range (1,101):
        if i%2 ==0:
            print(i,end=" ")
even_numbers()            
```
